```python
class OddEven():
    def OddEven():
        number=int(input("Enter a number:"))
        if((number%2)==0):
            message=str(number) + " Is Even"
        else:
            message=str(number) + " Is Odd"
        return message
```

```python
message=OddEven.OddEven()
print(message)
```

```python
class EligiblityForMarriage():
    def Eligible():
        age=int(input("Enter Your age:"))
        gender=input("Enter Your Gender:")
        if(gender.lower()=="female"):
            if(age>=18):
              eligible="You are eligible"
            else:
                eligible="You are not eligible"
        elif(gender.lower()=="male"):
            if(age>=21):
                eligible="You are eligible"
            else:
                eligible="You are not eligible"
        return eligible
```

```python
eligible=EligiblityForMarriage.Eligible()
print(eligible)
```

```python
class FindPercent():
    def Percentage():
        tamil=float(input("Tamil:"))
        english=float(input("English:"))
        maths=float(input("Maths:"))
        science=float(input("Science:"))
        socialScience=float(input("SocialScience:"))
        total=tamil+english+maths+science+socialScience
        percentage=float(total/5)
        print("Total:",total)
        print("Percentage:",percentage)
```
