```python
import pickle
import pandas as pd
```

```python
dataset=pd.read_csv("C:\AI\Artificial-Intelligence\Machine Learning\Assignment_MachineLearning\insurance_pre.csv")
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
independant=dataset[["age","bmi","children","sex_male","smoker_yes"]]
dependant=dataset[["charges"]]
```
