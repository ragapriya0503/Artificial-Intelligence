```python
import pandas as pd
dataset=pd.read_csv("insurance_pre.csv")
dataset
```

```python
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
dataset
```

```python
independant=dataset[["age","bmi","children","sex_male","smoker_yes"]]
dependant=dataset[["charges"]]
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
param_grid={'criterion':['friedman_mse','absolute_error','poisson','squared_error'],'max_features':['sqrt','log2'],'n_estimators':[10,100]}
grid=GridSearchCV(RandomForestRegressor(),param_grid,refit=True,verbose=3,n_jobs=-1)
grid.fit(independant,independant)
```

```python
result=grid.cv_results_
```

```python
table=pd.DataFrame.from_dict(result)
table
```

```python
age=float(input("Age"))
bmi=float(input("BMI"))
child=float(input("Children"))
male=float(input("Male or Female ( 0 or 1)"))
smoke=float(input("Smoker yes or no ( 0 or 1)"))
```

```python
future_prediction=grid.predict([[age,bmi,child,male,smoke]])
future_prediction
```
