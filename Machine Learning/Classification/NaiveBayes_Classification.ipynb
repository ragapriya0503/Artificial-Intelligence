```python
import pandas as pd
dataset=pd.read_csv("Social_Network_Ads.csv")
dataset
```

```python
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
dataset
```

```python
dataset=dataset.drop('User ID',axis=1)
dataset
```

```python
dataset['Purchased'].value_counts()
```

```python
independant=dataset[['Age','EstimatedSalary','Gender_Male']]
dependant=dataset['Purchased']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(independant,dependant,test_size=0.30,random_state=0)
```

```python
#https://scikit-learn.org/stable/modules/naive_bayes.html
from sklearn.naive_bayes import GaussianNB
gn=GaussianNB()
gn.fit(X_train,Y_train)
Y_pred=gn.predict(X_test)

from sklearn.metrics import confusion_matrix
cm=confusion_matrix(Y_test,Y_pred)
print(cm)

from sklearn.metrics import classification_report
classy=classification_report(Y_test,Y_pred)
print(classy)
```

```python
#https://scikit-learn.org/stable/modules/generated/sklearn.naive_bayes.MultinomialNB.html#sklearn.naive_bayes.MultinomialNB
from sklearn.naive_bayes import MultinomialNB
mul=MultinomialNB()
mul.fit(X_train,Y_train)
Y_pred=mul.predict(X_test)

from sklearn.metrics import confusion_matrix
cm1=confusion_matrix(Y_test,Y_pred)
print(cm1)

from sklearn.metrics import classification_report
classy1=classification_report(Y_test,Y_pred)
print(classy1)
```

```python
#https://scikit-learn.org/stable/modules/generated/sklearn.naive_bayes.ComplementNB.html#sklearn.naive_bayes.ComplementNB
from sklearn.naive_bayes import ComplementNB
com=ComplementNB()
com.fit(X_train,Y_train)
Y_pred=com.predict(X_test)

from sklearn.metrics import confusion_matrix
cm2=confusion_matrix(Y_test,Y_pred)
print(cm2)

from sklearn.metrics import classification_report
classy2=classification_report(Y_test,Y_pred)
print(classy2)
```

```python
#https://scikit-learn.org/stable/modules/generated/sklearn.naive_bayes.BernoulliNB.html#sklearn.naive_bayes.BernoulliNB
from sklearn.naive_bayes import BernoulliNB
br=BernoulliNB()
br.fit(X_train,Y_train)
Y_pred=br.predict(X_test)

from sklearn.metrics import confusion_matrix
cm3=confusion_matrix(Y_test,Y_pred)
print(cm3)

from sklearn.metrics import classification_report
classy3=classification_report(Y_test,Y_pred)
print(classy3)
```

```python
#https://scikit-learn.org/stable/modules/generated/sklearn.naive_bayes.CategoricalNB.html#sklearn.naive_bayes.CategoricalNB
from sklearn.naive_bayes import CategoricalNB
cat=CategoricalNB()
cat.fit(X_train,Y_train)
Y_pred=cat.predict(X_test)

from sklearn.metrics import confusion_matrix
cm4=confusion_matrix(Y_test,Y_pred)
print(cm4)

from sklearn.metrics import classification_report
classy4=classification_report(Y_test,Y_pred)
print(classy4)
```
