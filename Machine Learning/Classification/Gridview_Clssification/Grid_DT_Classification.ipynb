```python
import pandas as pd
dt=pd.read_csv('C:\AI\Artificial-Intelligence\Machine Learning\Classification\Social_Network_Ads.csv')
dt
```

```python
dt=pd.get_dummies(dt,dtype=int,drop_first=True)
dt
```

```python
dt=dt.drop('User ID',axis=1)
dt
```

```python
ip=dt[['Age','EstimatedSalary','Gender_Male']]
op=dt['Purchased']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(ip,op,test_size=0.30,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV
```

```python
grid_parm={'splitter':['best','random'],'criterion':['gini','entropy','log_loss'],'max_features':['sqrt','log2']}
grid_model=GridSearchCV(DecisionTreeClassifier(),grid_parm,refit=True,verbose=3,n_jobs=-1,scoring='f1_weighted')
grid_model.fit(x_train,y_train)
```

```python
result=grid_model.cv_results_
```

```python
grid_pred=grid_model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,grid_pred)
print(cm)
```

```python
from sklearn.metrics import classification_report
classy=classification_report(y_test,grid_pred)
print(classy)
```

```python
from sklearn.metrics import roc_auc_score # r-receiver, o-operating, c- curve | a-area, u- under, c-curve
roc_auc_score(y_test,grid_model.predict_proba(x_test)[:,1])
```
