```python
text=input("Enter the string")
text=text.replace(" ","")
reverseString=text[::-1]
if(text.lower()==reverseString.lower()):
    print("Given string is Anagram")
else:
    print("Given string is not an Anagram")
    
```

```python
minBalance=1000
def CheckBalance():
    amount=float(input("Enter the amount to withdraw "))
    if(amount>minBalance):
        print("Withdraw Processing")
    elif(amount<=minBalance):
        print("Insufficient Balance")

CheckBalance()        
```

```python
lists=[12,45,67,89,67,90,76,12]
duplicateNumbers=[]
uniqueNumbers=[]
for x in lists:
    if(x not in uniqueNumbers):
        uniqueNumbers.append(x)
    else:
        duplicateNumbers.append(x)
        
print("Unique Numbers: ",uniqueNumbers) 
print("Duplicate Numbers: ",duplicateNumbers) 
```
