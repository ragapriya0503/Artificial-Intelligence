```python
def factorial(number):
    if number == 0:
        return 1  
    return number * factorial(number - 1) 
```

```python
outNumber=print(factorial(5))
```

```python
number=int(input("Enter the number"))
lengthOftheNumber=len(str(number))
result=0
if(number>0):
    for x in str(number):
        result=result+int(x)**int(lengthOftheNumber)
if(result==number):
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
import re
sentence = input("Enter a sentence: ")
words = re.split(r'[ .*]', sentence)
longestWord = ""
#print(words)
for word in words:
    if len(word) > len(longestWord):
        longestWord = word

print("The longest word is: " ,longestWord)
```
