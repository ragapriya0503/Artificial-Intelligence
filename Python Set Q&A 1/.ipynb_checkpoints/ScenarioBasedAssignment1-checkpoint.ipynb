```python
salary=float(input("Enter your salary"))
if(salary>=50000):
     bonus=float(salary*(10/100))
     finalsalary=salary+bonus
     print(finalsalary)
else:
     print(salary)        
```

```python
number=int(input("Enter the number"))
if((number%2==0)):
    print(str(number) + " Is Even")
else:
    print(str(number) + " Is Odd")
```

```python
tamil=float(input("Enter Tamil mark: "))
english=float(input("Enter English mark: "))
maths=float(input("Enter Maths mark: "))
science=float(input("Enter Science mark: "))
socialScience=float(input("Enter Social Science mark: "))
total= tamil+english+maths+science+socialScience
if(tamil>=40 and english>=40 and maths>=40 and science>=40 and socialScience>=40):
    print("You are Pass")
    print(total)
else:
    print("You are Fail")
    print(total)      
```

```python
purchaseAmount=float(input("Enter the purchased amount:"))
if(purchaseAmount>=100):
    discount=float(purchaseAmount*(20/100))
    print("You can pay:",purchaseAmount-discount)
    print("Discount:",discount)
else:
    print(purchaseAmount)
```

```python
def withdrawal(balance:float):
    if(balance>=5000):
        print("Balance: ",balance)
        print("Withdrawal Processing")
    else:
        print("Balance: ",balance)
        print("Balance is less than 5000.Not able to proceed")
    
```

```python
withdrawal(10000)
withdrawal(2000)
```

```python
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(str(year) + " is leap year")
else:
    print(str(year) + " is not a leap year")
```
